**2. Calcular rentabilidades mensuales**

Calculamos rentabilidad de las acciones con la formula

Captura de pantalla 2025-05-18 174808.png

```python
print(df.columns)
```

```python
df_numeric = df.select_dtypes(include='number')
rentabilidad = (df_numeric - df_numeric.shift(1)) / df_numeric.shift(1)
print(rentabilidad)
```

**3. Cálculos agregados sobre rentabilidades**

```python
# Rentabilidad promedio mensual.

# Calcular rentabilidad promedio por acción (columna)
promedio_por_accion = rentabilidad.mean(axis=0, skipna=True)

# Mostrar resultado
print(promedio_por_accion)
```
